No exercício 20, eu entendi que a matriz transposta troca as linhas pelas colunas.

```python
## Exercício 21

def busca_binaria(lista, valor):
  inicio = 0
  fim = len(lista) - 1

  while inicio <= fim:
    meio = (inicio + fim)  // 2

    if lista[meio] == valor:
      return meio
    elif lista[meio] < valor:
      inicio = meio + 1
    else:
      fim = meio - 1

  return -1

numeros = [1, 3, 5, 7, 9, 11]

resultado = busca_binaria(numeros, 7)

print("Posição:", resultado)
```

no exercício 21, eu entendi que a busca binária procura o elemento dividindo a lista pela metade a cada tentantiva. A lista precisa estar ordenada.

```python
## Exercício 22

def criptografar(texto, deslocamento):
  resultado = ""

  for caractere in texto:
    if caractere.isalpha():
      inicio = ord('A') if caractere.isupper() else ord('a')
      resultado += chr((ord(caractere) - inicio + deslocamento) % 26 + inicio)
    else:
      resultado += caractere

  return resultado

mensagem = input("Digite a mensagem:")
chave = int(input("Digite o deslocamento:"))

print(criptografar(mensagem, chave))
```

No exercício 22, eu entendi que a Cifra de César troca cada letra por outra algumas posições à frente no alfabeto.

```python
## Exercício 23

def descriptografar(texto, deslocamento):
  resultado = ""

  for caractere in texto:
    if caractere.isalpha():
      inicio = ord('A') if caractere.isupper() else ord('a')
      resultado += chr((ord(caractere) - inicio - deslocamento) % 26 + inicio)
    else:
      resultado += caractere

  return resultado

mensagem = input("Digite a mensagem criptografada:")
chave = int(input("Digite o deslocamento:"))

print(descriptografar(mensagem, chave))
```

No exercício 23, eu fiz o processo contrário da criptografia, diminuindo o deslocamento para recuperar as letras originais.

```python
## Exercício 24

def validar_cpf(cpf):
  cpf = ''.join(c for c in cpf if c.isdigit())

  if len(cpf) != 11:
    return False

  if cpf == cpf[0] * 11:
    return False

  soma = sum(int(cpf[i]) * (10 - i) for i in range(9))
  resto = (soma * 10) % 11

  if resto == 10:
    resto = 0

  if resto != int(cpf[9]):
    return False

  soma = sum(int(cpf[i]) * (11 - i) for i in range(10))
  resto = (soma * 10) % 11

  if resto == 10:
    resto = 0

  return resto == int(cpf[10])

cpf = input("Digite o CPF:")

if validar_cpf(cpf):
  print("CPF válido.")
else:
  print("CPF inválido.")
```

No exercício 31, eu entendi que o Bubble Sort compara números vizinhos e troca suas posições quando estão na ordem errada.

```python
## Exercício 32

tabuleiro = [" "] * 9

def mostrar():
    print(tabuleiro[0], "|", tabuleiro[1], "|", tabuleiro[2])
    print("--+---+--")
    print(tabuleiro[3], "|", tabuleiro[4], "|", tabuleiro[5])
    print("--+---+--")
    print(tabuleiro[6], "|", tabuleiro[7], "|", tabuleiro[8])

def venceu(jogador):
    combinacoes = [
        (0,1,2), (3,4,5), (6,7,8),
        (0,3,6), (1,4,7), (2,5,8),
        (0,4,8), (2,4,6)
    ]

    return any(
        tabuleiro[a] == tabuleiro[b] == tabuleiro[c] == jogador
        for a, b, c in combinacoes
    )

jogador = "X"

for rodada in range(9):
    mostrar()

    posicao = int(input(f"Jogador {jogador}, escolha uma posição (1-9): ")) - 1

    if tabuleiro[posicao] != " ":
        print("Posição ocupada!")
        continue

    tabuleiro[posicao] = jogador

    if venceu(jogador):
        mostrar()
        print("Jogador", jogador, "venceu!")
        break

    jogador = "O" if jogador == "X" else "X"
else:
    mostrar()
    print("Empate!")
```

```python
## Exercício 48

numeros = [5, 2, 4, 6, 1, 3]

for i in range(1, len(numeros)):
    chave = numeros[i]
    j = i - 1

    while j >= 0 and numeros[j] > chave:
        numeros[j + 1] = numeros[j]
        j -= 1

    numeros[j + 1] = chave

print(numeros)
```
